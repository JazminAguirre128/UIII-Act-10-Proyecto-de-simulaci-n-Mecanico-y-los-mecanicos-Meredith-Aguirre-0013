## Mecanico

```python
losmecanicos = []
mecanico = {}

while True:
    print("\n--- MENÚ IA MECÁNICOS ---")
    print("1. Registrar Mecánico")
    print("2. Ver Lista y Diagnóstico IA")
    print("3. Salir")

    opcion = input("Elige una opción: ")

    if opcion == "1":
        nombre = input("Nombre: ")
        experiencia = int(input("Años de experiencia: "))
        autos = int(input("Autos reparados este mes: "))

        mecanico = {
            "nombre": nombre,
            "experiencia": experiencia,
            "autos": autos,
            "rango_ia": "No calculado"
        }
        losmecanicos.append(mecanico)
        print("✅ Mecánico guardado.")

    elif opcion == "2":
        if losmecanicos == []:
            print("📭 La lista está vacía.")
        else:
            print("\n--- EVALUACIÓN CON IA ---")
            for m in losmecanicos:
                if m["experiencia"] >= 5 and m["autos"] >= 15:
                    m["rango_ia"] = "Experto (Fallas difíciles)"
                elif m["experiencia"] >= 2:
                    m["rango_ia"] = "Intermedio (Motores)"
                else:
                    m["rango_ia"] = "Principiante (Mantenimiento básico)"

                print(f"• {m['nombre']} -> Experiencia: {m['experiencia']} años | Autos: {m['autos']} | IA: {m['rango_ia']}")

    elif opcion == "3":
        print("🤖 Programa terminado. ¡Adiós!")
        print("Programa realizado por Meredith Aguirre NL = 2")
        break

    else:
        print("❌ Opción incorrecta.")
```
